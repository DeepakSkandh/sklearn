### Another way to reduce no of features is through unsupervised dimensionality techniques

```python
from sklearn.decomposition import PCA
```

```python
from sklearn.datasets import load_diabetes

data = load_diabetes(as_frame=True)

X = data.data
y = data.target
```

```python
data.keys()
```

```python
X.shape
```

```python
histogram = X.hist(figsize=(12,8)) #histogram of features
```

```python
import matplotlib.pyplot as plt

X.plot(kind="box", figsize=(15,5))
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled.mean(axis=0)
```

```python
X_scaled.std(axis=0)
```

### PCA on Data

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
X_pca.shape
```

### Transformed data

```python
import pandas as pd

pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

pca_df.head()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"]
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Projection")

plt.show()
```
