### NUMERICAL TRANSFORMATION

####  Standard Scalar

$$
z = \frac{x - \mu}{\sigma}
$$

```python
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
x = np.array([
    [10],[20],[3],[4],[100],[63]
])
x
```

```python
ss = StandardScaler()
x_new = ss.fit_transform(x)
x_new
```

### Mean is zero and Standard Deviation is 1 after performing Scaling

```python
print(int(x_new.mean()))
print(round(x.mean(),2))
```

```python
print(round(x.std(),2))
print(x_new.std())
```

### MinMax Scalar

Range = [0,1]

$$
x_{\text{scaled}} = \frac{x - x_{\min}}{x_{\max} - x_{\min}}
$$

Which means transformed minimum x is 0 and maximum x is 1

```python
from sklearn.preprocessing import MinMaxScaler

x = np.array([
    [10],[20],[30],[45],[34]
])
mm_scale = MinMaxScaler()
x_new = mm_scale.fit_transform(x)
print(x_new)
```

### They are sensitive to Outliers

```python
x = np.array([
    [10],[20],[30],[40],[1000]
])
mm_scale = MinMaxScaler()
x_new = mm_scale.fit_transform(x)
print(x_new) # the outlier 1000 squeezed others to zero
```

### MaxAbsScalar

$$
x_{\text{scaled}}=\frac{x}{|x|_{\max}}
$$

```python
from sklearn.preprocessing import MaxAbsScaler

x = np.array([[10],
              [20],
              [30],
              [40],
              [-50]])

scaler = MaxAbsScaler()

x_scaled = scaler.fit_transform(x)

print(x_scaled)
```

### Functional Transformer

Constructs transformed features by applying user defined function

```python
from sklearn.preprocessing import FunctionTransformer
```

```python
x = np.array([
    [4],[16],[20],[10]
])
func_transform = FunctionTransformer(np.log2)
x_new = func_transform.fit_transform(x)
x_new
```

```python
x = np.array([
    [4],[16],[20],[10]
])
func_transform = FunctionTransformer(np.sqrt)
x_new = func_transform.fit_transform(x)
x_new
```

## Polynomial transformations

### Does explicit feature Uplifting

#### Original Feature Vector

$$
X = \begin{bmatrix}
x_1 & x_2
\end{bmatrix}
$$

#### After Polynomial Transformation (Degree = 2)

$$
\phi(X)=
\begin{bmatrix}
1,\;
x_1,\;
x_2,\;
x_1^2,\;
x_1x_2,\;
x_2^2
\end{bmatrix}
$$

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
x = np.array([
    [4,10],[2,1],[5,6],[9,0]
])

uplift = PolynomialFeatures(degree=2)
x_new = uplift.fit_transform(x)
x_new
```

```python
uplift = PolynomialFeatures(degree=3)
x_new = uplift.fit_transform(x)
x_new
```
