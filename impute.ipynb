## Imputation in sklearn

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 25, np.nan, 35, 40],
    "Salary": [30000, np.nan, 50000, 60000, 70000]
})

df
```

```python
from sklearn.impute import SimpleImputer
```

#### By default fit_transform() converts DataFrame into Numpy array

```python
imputer1 = SimpleImputer(strategy="mean")
x1 = imputer1.fit_transform(df)
x1
```

#### Use Mean when there are no strong outliers, use Median when there is a skewed distribution

```python
imputer2 = SimpleImputer(strategy="median")

x2 = imputer2.fit_transform(df)
x2
```

```python
imputer3 = SimpleImputer(strategy="constant",fill_value=0)
x3 = imputer3.fit_transform(df)
x3
```

#### One imputer with strategy="constant" uses the same fill_value for all columns it transforms.

#### individual columns are handled by individual imputers

```python
age_imputer = SimpleImputer(strategy="constant", fill_value=0)
salary_imputer = SimpleImputer(strategy="constant", fill_value=10000)

df[["Age"]] = age_imputer.fit_transform(df[["Age"]])
df[["Salary"]] = salary_imputer.fit_transform(df[["Salary"]])
```

## KNN IMPUTER

```python
df = pd.DataFrame({
    "Age":    [20, 22, 40, 42, 21],
    "Salary": [30000, 32000, 80000, 85000, np.nan]
})
df
```

```python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=2)

X = imputer.fit_transform(df)

X
```
