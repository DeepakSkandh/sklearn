### Ordinal Encoder


```python
from sklearn.preprocessing import OrdinalEncoder
```

#### Sorts Alphabetically not Logically

```python
import pandas as pd

df = pd.DataFrame({
    "Size" : ["Medium","Large","Small","Medium"]
})
df
```

```python
encoder = OrdinalEncoder()
x = encoder.fit_transform(df)
x
```

```python
encoder.categories_
```

### To specify the order

```python
encoder1 = OrdinalEncoder(
    categories=[["Small","Medium","Large"]]
)
```

```python
x1 = encoder1.fit_transform(df)
x1
```

## ONE HOT ENCODER

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np
```

```python
x = np.array([[1],[2]])
new = OneHotEncoder()
x1 = new.fit_transform(x)
x1
```

```python
x = np.array([[1],[2],[3],[1]])
new = OneHotEncoder(sparse_output=False)
x1 = new.fit_transform(x)
x1
```

```python
df = pd.DataFrame({
    "age" : [10,20,10,43,34],
    "Name" : ["Rahul","Suresh","migul","sekar","raja"]
})
df
```

```python
encoder = OneHotEncoder(sparse_output=False)
x = encoder.fit_transform(df)
x
```

```python
encoder.get_feature_names_out()
```

```python
x = encoder.fit_transform(df[['Name']])
x
```

### Binary classification task

```python
y = ["Yes", "No", "Yes", "No"]
```

```python
from sklearn.preprocessing import LabelBinarizer

lb = LabelBinarizer()

y1 = lb.fit_transform(y)

print(y1)
```

```python
y = ["Cat", "Dog", "Bird", "Dog"]
```

```python
lb = LabelBinarizer()

print(lb.fit_transform(y))
```
